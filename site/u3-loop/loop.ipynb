### 1-3. 프로그램 실행 순서 (137쪽 스스로 해결하기)

### 1-5. while문 · 만세 (140쪽)
`3`을 다른 수로 바꿔 실행해 봐요. 마지막 줄 `i += 1`이 없으면 어떻게 될지 생각해 보고, 바꿔서 실행했다면 ■ 버튼으로 멈춰요.

```python
print('대한민국')
i = 0
while i < 3:
    print('만세!')
    i += 1
```

### 1-6. break와 continue (141쪽)
화면 05에서 출력을 먼저 예측한 뒤, 두 셀을 차례로 실행해 예측과 비교해요.

```python
score = [10, 20, -10, 30, -40, 20]
for i in score:
  if i < 0:
    break
  print(i)
print('반복문 종료')
```

```python
score = [10, 20, -10, 30, -40, 20]
for i in score:
  if i < 0:
    continue
  print(i)
print('반복문 종료')
```
