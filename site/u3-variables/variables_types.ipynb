## 1. 교과서 예제 따라 하기

### 1-1 type() 함수로 자료형 확인하기 (120쪽)
파이썬은 저장되는 데이터에 맞게 **자동으로 자료형을 지정**해요. `type()` 함수로 변수에 저장된 자료형을 확인할 수 있어요.

````python
a = 100
b = 3.1415
c = 'Python'
d = True
print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))
````

여러 개의 값을 담는 **컨테이너형**도 확인해요. 괄호 모양을 잘 봐요.

````python
colors = ['빨', '주', '노', '초', '파']
weekday = ('월', '화', '수', '목', '금')
students = {'혜민':'131-9898', '서준':'457-4777'}
print(type(colors), type(weekday), type(students))
````

### 1-2 변수는 한 번에 한 개의 값만 저장해요 (120쪽)
이미 값이 있는 변수에 새 값을 저장하면 기존 값은 지워지고, **자료형도 새 값에 맞게 바뀌어요**.

````python
a = 100
print(a, '을 저장했을 때 a의 자료형', type(a))
a = 'Python'
print(a, '을 저장했을 때 a의 자료형', type(a))
````

### 3-2 (선택) 프로그램을 실행한 결과를 확인하고 까닭 이야기하기
3-3을 먼저 하고, 시간이 남으면 해요. 실행하기 전에 **area에 저장될 값과 자료형**을 예측하고, 실행한 뒤 까닭을 화면 04의 선택 메모에 적어요.

> 💡 소수점이 있으면 실수형이 되고, 작은따옴표나 큰따옴표로 감싸면 문자열이 돼요.
> 📌 `2 * pi * radius`는 변수 이름이 area(넓이)이지만 실제로는 **원의 둘레**를 구하는 식이에요(원의 넓이는 `pi * radius * radius`). 자료형을 확인하는 데에는 상관없어요.

````python
radius = 5
pi = 3.141592
area = 2 * pi * radius
print('area에 계산된 값은', area, '이고 자료형은', type(area), '이다.')
````

### 3-3 서로 다른 자료형을 연산하면?
`a, b, c = 30, 30.0, '30'`은 a, b, c에 값을 차례로 하나씩 저장하는 코드예요. 실행하기 전에 **sum1과 sum2가 어떻게 될지** 예측해요.

> 💡 **TypeError**: 연산이나 함수에 전달되는 자료형이 잘못된 경우에 발생하는 에러예요. 오류 메시지의 마지막 줄을 읽어 봐요.

````python
a, b, c = 30, 30.0, '30'
print(a, b, c)
print(type(a), type(b), type(c))

sum1 = a + b
print(sum1)

sum2 = a + c
print(sum2)
````
