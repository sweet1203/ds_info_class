### 1-1 표준 출력 print() 함수 (122쪽)

실행하면 아래에 입력 상자가 나와요. 나이를 입력하고 Enter를 눌러요.

### 1-3 파일 쓰기와 파일 읽기 (123쪽)
교과서 123쪽 코드와 같은 구조예요. 파일 이름과 내용만 바꿨어요. 내용을 **나의 다짐**으로 바꿔 써도 좋아요.

- `open(파일명, 'w')` → `write()` → `close()` : 파일 열기 → 파일 처리(쓰기) → 파일 닫기
- `\n`은 줄을 바꾸라는 표시예요. `#` 뒤의 글은 **주석**이라 실행되지 않아요.

실행한 뒤 왼쪽 **📁** 에서 `나의_다짐.txt`가 생겼는지 확인해요.

````python
# 파일 쓰기
f = open('나의_다짐.txt', 'w')
f.write('하루에 물 여덟 잔 마시기\n')
f.write('잠들기 전 휴대폰 내려놓기\n')
f.close()
````

````python
# 파일 읽기
f = open('나의_다짐.txt', 'r')
a = f.read()
print(a)
f.close()
````

### 1-4 'a' 모드로 이어 쓰고 readlines()로 읽기
- `'a'`(추가) 모드는 파일에 있던 데이터를 그대로 두고 **이어 써요**.
- `readlines()`는 모든 줄을 읽어서 **각 줄을 요소로 갖는 리스트**로 돌려줘요.

이 칸을 **한 번 더 실행**하면 어떻게 될지 예측한 뒤 실행해 봐요. `'a'`를 `'w'`로 바꾸면 어떻게 될까요?

````python
f = open('나의_다짐.txt', 'a')
f.write('하루에 한 번 하늘 보기\n')
f.close()

f = open('나의_다짐.txt', 'r')
lines = f.readlines()
print(lines)
f.close()
````
