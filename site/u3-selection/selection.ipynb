## 1. 교과서 예제 실행해 보기

### 1-1. if문 (133쪽)
'하차'일 때 문이 열리고 다음 정거장을 알리는 프로그램이에요. **y**와 **n**을 넣어 두 번 실행하고, 출력이 어떻게 달라지는지 비교해요.

```python
print('이번 정거장은 서울역입니다.')
bus = input('하차하시겠습니까?(y/n)')
if bus == 'y':
  print('문이 열립니다. 하차하세요.')
print('다음 정거장은 시청역입니다.')
```

### 1-2. if~else문 (133쪽)
짝수인지 홀수인지를 판단하는 프로그램이에요. 짝수와 홀수를 하나씩 넣어 실행해요.
`%` 연산자는 나눗셈의 나머지를 계산해요.

```python
number = int(input('숫자를 입력하세요.'))
if number % 2 == 0:
  print(number,'는 짝수입니다.')
else:
  print(number,'는 홀수입니다.')
```

### 1-3. if~elif~else문 (134쪽)
숫자를 입력받아 양수, 음수, 0을 판단하는 프로그램이에요. **양수, 음수, 0**을 하나씩 넣어 세 번 실행해요.
`end = ''` 때문에 결과가 어떻게 출력되는지도 살펴봐요.

```python
number = int(input('숫자를 입력하세요: '))
if number > 0:
  print('양수', end = '')
elif number < 0:
  print('음수', end = '')
else:
  print('0', end = '')
print('입니다.')
```
