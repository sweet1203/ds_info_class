### 1-1. 무한 루프 · 곱셈 문제 (142쪽)
정답을 맞힐 때까지 반복하여 곱셈 문제를 내는 프로그램이에요. 일부러 한 번 틀려 보고, **난수의 범위를 수정해서** 다시 실행해 봐요.

```python
import random
while True:
  a = random.randint(1, 9)
  b = random.randint(1, 9)
  print(a, '*', b, '= ?')
  num = int(input('정답 입력:'))

  if num == a * b:
    print('정답입니다!')
    break

  print(f'틀렸습니다. 정답은 {a * b}입니다.')
```

### 1-3. 함수 정의와 함수 호출 (144쪽)
1부터 입력받은 수까지 누적 합을 구하는 함수를 만드는 프로그램이에요. 10을 넣어 실행해 봐요.

```python
def total(number):
    sum = 0
    for i in range(1, number + 1):
      sum +=i
    return sum
num = int(input('수: '))
print(total(num))
```

### 1-4. 지역 변수와 전역 변수 (144쪽)
10까지의 약수 개수와 20까지의 약수 개수를 합하는 프로그램이에요. 함수 안의 `tot`(지역 변수)과 함수 밖의 `tot`(전역 변수)이 각각 어떤 값을 가지는지 출력을 보며 생각해 봐요.

```python
def divisorCount(n):
  tot = 0
  for i in range(1, n+1):
    if n % i == 0:
      tot += 1
  print('약수 개수:', tot)
  return tot

tot = 0
tot += divisorCount(10)
tot += divisorCount(20)
print('약수 누적 개수:', tot)
```
