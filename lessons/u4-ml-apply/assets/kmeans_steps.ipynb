## 1. 라이브러리 불러오기 (따라 하기)
- `matplotlib.pyplot`: 그래프와 시각화를 위한 모듈
- `make_blobs`: 군집화 연습용 데이터를 만드는 함수
- `KMeans`: k-평균 알고리즘을 구현한 클래스

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

## 2. 데이터 생성 (따라 하기)
`make_blobs`로 점 20개를 만들어요. `random_state=4`로 정하면 실행할 때마다 같은 데이터가 만들어져요.
레이블(정답)은 쓰지 않으므로 `_`로 받아 버려요. 비지도학습이니까요!

```python
X, _ = make_blobs(n_samples=20, random_state=4)

plt.figure(figsize=(4, 4))
plt.scatter(X[:, 0], X[:, 1], c='gray')
plt.title('data (no label)')
plt.show()
```

## 2-1. k-평균을 끝까지 한 번 실행해 보기 (따라 하기)
군집의 개수를 2로 정하고 학습해요. 학습이 끝나면
- `model.labels_`: 점마다 어느 군집(0 또는 1)에 들어갔는지
- `model.cluster_centers_`: 군집마다 중심의 좌표

를 알 수 있어요. 점은 원(`'o'`), 중심은 큰 사각형(`'s'`)으로 그려요.

```python
model = KMeans(n_clusters=2, n_init=10, random_state=0).fit(X)   # 학습
print('군집 번호:', model.labels_)

colors = ['r', 'b']
plt.figure(figsize=(4, 4))
for idx, color in zip(range(2), colors):
    plt.scatter(X[model.labels_ == idx, 0], X[model.labels_ == idx, 1], marker='o', facecolor=color)
    plt.scatter(model.cluster_centers_[idx, 0], model.cluster_centers_[idx, 1], marker='s', c=color, s=200)
plt.title('k-means result (k=2)')
plt.show()
```
