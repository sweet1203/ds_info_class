## 1. 행성 데이터 준비하기 (156쪽)
행성 이름을 **키**로, 값을 **값**으로 하는 딕셔너리 두 개예요(126~128쪽 딕셔너리).

````python
# 행성의 공전 주기(단위: 일)
orbital_periods = {'수성': 88, '금성': 225, '지구': 365, '화성': 687, '목성': 4380, '토성': 10767, '천왕성': 30660, '해왕성': 60152}

# 행성으로부터 태양까지의 평균 거리(단위: km)
orbital_distances = {'수성': 57910000, '금성': 108200000, '지구': 149600000, '화성': 227900000, '목성': 778600000, '토성': 1433500000, '천왕성': 2872500000, '해왕성': 4495100000}

# 데이터 탐색: 행성마다 공전 주기와 평균 거리 보기
for planet in orbital_periods:
    print(planet, orbital_periods[planet], '일', orbital_distances[planet], 'km')
````
