## 1. 교과서 예제 따라 하기

### 1-1 리스트로 표현한 1차원 데이터 (127쪽)
`리스트 = [값 1, 값 2, 값 3]` — 대괄호 안에 값을 콤마로 구분해 저장해요.
- 값마다 **0번부터** 1씩 커지는 **인덱스**가 붙어요. 1번 학생의 데이터는 `T_size[0]`이에요.
- `append()`는 리스트에 값을 추가해요. 한 번에 값 하나씩 넣어요.

````python
# 리스트에 1차원 데이터 표현하기
T_size = [95, 85, 100, 95, 90]
print('학생의 상의 치수:', T_size)
# 인덱스를 이용해 리스트의 값 가져오기
print('1번 학생의 상의 치수:', T_size[0])
print('2번 학생의 상의 치수:', T_size[1])
# 인덱스를 이용해 리스트의 값 수정하기
T_size[0] = 100
# 1번 학생의 수정된 상의 치수 출력하기
print('1번 학생의 수정된 상의 치수:', T_size[0])
# 리스트에 값 추가하기
T_size.append(95)
T_size.append(105)
print(T_size)
````

### 1-2 딕셔너리로 표현한 1차원 데이터 (128쪽)
`딕셔너리 = {키1: 값1}`, `딕셔너리명[키]` — 중괄호 안에 **키:값** 쌍을 저장하고, 키로 값을 찾아요.
- 대괄호 안에 키를 넣어 값을 찾거나, 새로운 키:값을 저장할 수 있어요.
- 중복된 키는 쓸 수 없고, 이미 있는 키로 저장하면 기존 키의 값이 수정돼요.

````python
# 딕셔너리에 1차원 데이터 표현하기
T_size = {'내빈':95, '미송':85, '민수':100, '준서':95, '지환':90 }
print('우리 반 모든 학생의 상의 치수:', T_size)
# 키를 이용해 딕셔너리의 값 가져오기
print('내빈이의 상의 치수:', T_size['내빈'])
print('미송이의 상의 치수:', T_size['미송'])
# 키를 이용해 딕셔너리의 값 수정하기
T_size['내빈'] = 100 # 내빈이의 상의 치수를 100으로 수정
print(T_size)
# 딕셔너리에 새로운 키:값 쌍을 저장하기
T_size['재우'] = 105
print('우리 반 모든 학생의 상의 치수:', T_size)
````
