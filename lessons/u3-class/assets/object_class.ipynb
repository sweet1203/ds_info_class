### 1-3. 클래스 안에 메소드 정의하기 (149쪽)
`get_years_since_publication()` 메소드는 올해 연도에서 그 책의 출판 연도(`self.publish_year`)를 빼서 반환해요.
교과서는 2024년 기준이라 105가 나오지만, 올해 실행하면 결과가 달라져요.

```python
from datetime import date   # date: 날짜를 처리하는 표준 라이브러리 객체

class Book:
    def __init__(self, title, author, publish_year):
        self.title = title
        self.author = author
        self.publish_year = publish_year

    def get_years_since_publication(self):   # 책이 출판된 지 몇 년이 지났는지를 반환하는 메소드
        current_year = date.today().year
        return current_year - self.publish_year

book = Book('데미안', '헤르만 헤세', 1919)
print(book.get_years_since_publication())
```

### 1-4. 객체지향 프로그래밍 예제 · 학급 문고 대여 (149쪽)
Book 클래스는 도서명(title), 저자(author), 대여 여부(is_rented)를, Student 클래스는 학번(id), 이름(name), 대여한 책 리스트(rented_books)를 속성으로 가져요.
`rent_book()` 메소드는 책이 대여 중이 아니면 빌려주고, 대여 중이면 알려 줘요.

실행하기 **전에** 마지막 두 `print()`가 무엇을 출력할지 먼저 예측해 보세요.

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_rented = False

    def __str__(self):          # print(객체)를 쓰면 호출되는 스페셜 메소드
        return f'{self.title}'

class Student:
    def __init__(self, id, name):
        self.id = id
        self.name = name
        self.rented_books = []

    def rent_book(self, book):
        if not book.is_rented:
            book.is_rented = True
            self.rented_books.append(book)
            return f'{self.name}에게 {book.title} 대여함.'
        else:
            return f'{book.title} 대여 중.'

b1 = Book('소나기', '황순원')
s1 = Student(1, '홍길동')
print(s1.rent_book(b1))
s2 = Student(2, '이순신')
print(s2.rent_book(b1))
```

### 1-5. 프로그래밍 정리 · 런 길이 부호화 함수 (152쪽)
런 길이 부호화 방식(53쪽)으로 압축한 결과를 출력하는 함수예요. 함수 구조와, 프로그래밍으로 해결한 압축 방식을 살펴보세요.

```python
# 함수 정의
def run_length_encoding(text):
    a = text[0]
    cnt = 0
    for i in text:
        if i == a[-1]:
            cnt += 1
        else:
            a = a + str(cnt) + i
            cnt = 1
    a = a + str(cnt)
    print(a)
    return a

# 함수 호출
run_length_encoding('AAADDBOOOO')
```

**알고리즘 설계**

- 각 플레이어는 자기 순서에 자신의 숫자를 추측하고 그에 대한 피드백을 받는다.
- 모든 숫자를 정확하게 맞춘 플레이어가 있으면 승리한 플레이어의 이름을 출력하고 게임을 종료한다.

### 2-1. Player 클래스 (151쪽)
`random.sample(range(10), 3)`은 0~9 가운데 서로 다른 숫자 3개를 뽑아 리스트로 돌려줘요.

```python
import random

class Player:   # 클래스 정의
    def __init__(self, name=None):
        self.name = name
        self.numbers = random.sample(range(10), 3)   # 3개의 랜덤 야구 숫자 생성

    def check_strike(self, guessed_nums):   # strike와 ball 결과
        strike_count = ball_count = 0        # strike와 ball의 개수 변수를 0으로 초기화
        for i in range(3):
            if guessed_nums[i] == self.numbers[i]:   # 같은 위치의 숫자끼리 비교
                strike_count += 1                    # 숫자가 일치하면 strike 개수 증가
            elif guessed_nums[i] in self.numbers:    # 추측한 숫자가 정답에 있는지 검사
                ball_count += 1                      # 숫자가 존재하면 ball 개수 증가
        print(f'{strike_count} Strike, {ball_count} Ball')   # 판정 결과 출력
        return strike_count == 3   # strike_count가 3이면 True, 아니면 False를 반환함.
```

✏️ **check_strike() 시험하기** 정답을 직접 정해 두고 판정이 맞는지 확인해요. 실행하기 전에 결과를 먼저 예측해 보세요.

```python
p = Player('시험')
p.numbers = [4, 8, 6]          # 시험하려고 정답을 직접 정함
print(p.check_strike([4, 6, 8]))
print(p.check_strike([4, 8, 6]))
```

### 2-2. Game 클래스 · (가) 채우기 (151쪽)
`play()` 메소드 안의 **(가)** 부분을 채워 게임을 진행하게 만들어요. `check_gameover()`는 이미 완성되어 있어요.

**(가)에서 할 일** — 화면에서 늘어놓은 순서를 주석으로 적어 두었어요. 주석 아래에 코드를 써요.
1. 라운드 번호를 1로 정한다.
2. 게임이 끝날 때까지 아래 일을 되풀이한다.
3. 이번 라운드의 결과를 담을 빈 리스트(`game_result`)를 만든다.
4. 라운드 번호를 출력한다.
5. 참가자마다 숫자 3개를 입력받아 `check_strike()`로 판정하고, 그 결과를 리스트에 넣는다.
6. 라운드 번호를 1 늘린다.
7. `check_gameover()`로 끝났는지 확인하고, 끝났으면 반복을 멈춘다. ← 교과서에서 (가) 아래에 있는 두 줄이에요.

<details><summary>💡 힌트 1 · 어떤 반복문을 쓸까?</summary>

몇 라운드 만에 끝날지 미리 알 수 없어요. 횟수를 정하는 for 문보다, 조건을 늘 참으로 둔 while 문으로 되풀이하다가 끝나는 조건에서 break로 빠져나오는 방법이 알맞아요(136~141쪽 반복 구조).
</details>

<details><summary>💡 힌트 2 · 참가자마다 차례대로 하려면?</summary>

참가자 객체들은 Game 객체의 players 속성(리스트)에 들어 있어요. 이 리스트를 for 문으로 차례대로 꺼내면, 꺼낸 객체마다 이름 속성과 check_strike() 메소드를 쓸 수 있어요. 메소드 안에서 Game 객체 자신의 속성은 self로 가리켜요.
</details>

<details><summary>💡 힌트 3 · 「3 4 5」처럼 입력한 숫자 3개를 리스트로 바꾸려면?</summary>

입력받은 문자열을 split() 메소드로 띄어쓰기마다 나누면 문자열 3개가 돼요. 그런데 check_strike()는 정수끼리 비교하니 정수로 바꿔야 해요. 여러 값에 한꺼번에 같은 함수를 적용하는 map() 함수와, 그 결과를 리스트로 만드는 list() 함수를 떠올려 보세요(142쪽 map()·split() 함수).
</details>

<details><summary>💡 힌트 4 · 결과는 어떻게 모을까?</summary>

check_strike()는 3 Strike이면 True, 아니면 False를 반환해요. 참가자마다 돌려받은 값을 리스트 뒤에 하나씩 덧붙이는 메소드로 game_result에 모으면, check_gameover()가 그 리스트를 보고 끝낼지 정해요(126~128쪽 리스트).
</details>

<details><summary>💡 힌트 5 · 실행 예시처럼 출력하려면?</summary>

- 라운드 줄: 문자열에 정수를 곱하면 그 횟수만큼 되풀이돼요. 별표 15개 사이에 f-문자열로 라운드 번호를 넣어요.
- 차례 줄: print() 함수의 end 매개 변수를 빈 문자열로 주면 줄을 바꾸지 않아서, 이름 바로 뒤에서 숫자를 입력받을 수 있어요.
</details>

```python
class Game:   # 클래스 정의
    def __init__(self, players):   # players: 게임에 참여하는 Player 객체가 담긴 리스트
        self.players = players

    def play(self):
        # ───── (가) 여기부터 채워요 ─────
        # 1. 라운드 번호를 1로 정한다.

        # 2. 게임이 끝날 때까지 아래 일을 되풀이한다. (3~7은 반복 안쪽, 한 칸 더 들여 써요)

            # 3. 이번 라운드의 결과를 담을 빈 리스트 game_result를 만든다.

            # 4. 라운드 번호를 출력한다.

            # 5. 참가자마다 숫자 3개를 입력받아 check_strike()로 판정하고, 결과를 game_result에 넣는다.

            # 6. 라운드 번호를 1 늘린다.

        # ───── (가) 여기까지 ─────
            # 7. 결과 판별(모든 숫자를 맞힌 Player가 있으면 반복문을 빠져나감.)
            #    아래 두 줄은 맨 앞의 「# 」만 지우면 5번과 같은 깊이(반복 안쪽)가 돼요.
            # if self.check_gameover(game_result):
            #     break
        pass   # ← (가)를 다 쓰면 이 줄은 지워요.

    def check_gameover(self, game_result):   # 게임 종료 여부 판정
        if sum(game_result):   # 모든 숫자를 맞힌 True(1) 값이 하나라도 있으면 GAME OVER 처리
            print(f"\n\n{'<GAME OVER>':^40}"); print('=' * 40)
            for i in range(len(game_result)):
                if game_result[i]:   # 현재 Player의 game_result값이 True이면
                    print(f'{self.players[i].name} Win!')   # 해당 Player의 승리를 출력
            print('=' * 40)
            return True   # 모든 숫자를 맞힌 Player가 한 명이라도 있으면 True를 반환함.
```
