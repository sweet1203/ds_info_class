### 1-1 2차원 리스트로 의복 치수 표현 (129쪽)
`리스트명[행번호][열번호]` — 행과 열 모두 0부터 시작해요. 대괄호를 한 개만 쓰면 행 전체를 가리켜요.

````python
class1 = [[95, 250, 'M'],
          [85, 230, 'S'],
          [100, 270, 'XL'],
          [95, 235, 'L'],
          [90, 250, 'L']]

# 인덱스를 이용해 2차원 리스트의 값 가져오기
print('1번 학생의 의복 치수:', class1[0])
print('2번 학생의 상의 치수:', class1[1][0])
print('2번 학생의 신발 치수:', class1[1][1])

# 2차원 리스트의 값 수정하기
class1[4][0] = 100      # 5번 학생의 상의 치수 수정
class1[4][2] = 'XL'    # 5번 학생의 하의 치수 수정
print('5번 학생의 수정된 상의 치수:', class1[4][0])
print('5번 학생의 수정된 하의 치수:', class1[4][2])
````

### 1-2 2차원 딕셔너리로 의복 치수 표현 (130쪽)
`딕셔너리[키][키]` — 첫 번째 대괄호에는 바깥쪽 딕셔너리의 키, 두 번째 대괄호에는 내부 딕셔너리의 키를 넣어요.
(같은 이름 `class1`을 다시 쓰니, 이제부터 `class1`은 딕셔너리예요.)

````python
class1 = {
    '내빈': {'상의': 95, '신발': 250, '하의': 'M' },
    '미송': {'상의': 85, '신발': 230, '하의': 'S' },
    '민수': {'상의': 100, '신발': 270, '하의': 'XL' },
    '준서': {'상의': 95, '신발': 235, '하의': 'L' },
    '지환': {'상의': 90, '신발': 250, '하의': 'L' }
}

# 2차원 딕셔너리에서 키를 이용하여 값 가져오기
print('미송이의 의복 치수:', class1['미송'] )
print('민수의 상의 치수:', class1['민수']['상의'])

# 2차원 딕셔너리에서 키를 이용하여 값 수정하기
class1['준서']['하의'] = 'M'
print(class1['준서'])
````
